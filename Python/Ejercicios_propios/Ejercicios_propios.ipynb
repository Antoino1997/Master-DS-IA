### Ejercicio 1.1 – Estadísticas de temperaturas

Dispones de las siguientes temperaturas registradas durante una semana:

lunes = 21.5
martes = 23.0
miercoles = 19.8
jueves = 25.2
viernes = 22.7
sabado = 27.1
domingo = 24.3

Calcula:
1. Temperatura media.
2. Temperatura máxima y mínima.
3. Diferencia entre máxima y mínima.
4. Porcentaje de días con temperaturas superiores a 23 °C.
5. Determina si la temperatura media ha superado los 22 °C.

```python
lunes = 21.5
martes = 23.0
miercoles = 19.8
jueves = 25.2
viernes = 22.7
sabado = 27.1
domingo = 24.3

temperaturas = [lunes, martes, miercoles, jueves, viernes, sabado, domingo]
temperatura_media = (sum(temperaturas) / len(temperaturas))
diferencia_max_min = max(temperaturas) - min(temperaturas)
porcentaje_superior_23 = sum(1 for temp in temperaturas if temp > 23) / len(temperaturas) * 100

print("===== TEMPERATURAS =====")
print(f"Temperatura media de la semana: {temperatura_media:.2f} °C") # Temperatura media de la semana
print(f"Diferencia entre la temperatura más alta y la más baja: {diferencia_max_min:.2f} °C") # Diferencia entre la temperatura más alta y la más baja
print(f"Porcentaje de días con temperatura superior a 23 °C: {porcentaje_superior_23:.2f}%") # Porcentaje de días con temperatura superior a 23 °C
print(f"¿Ha superado la temperatura media los 22 °C?: {'Sí' if temperatura_media > 22 else 'No'}") # Indica si la temperatura media ha superado los 22 °C
```

### Ejercicio 1.2 – Rentabilidad de una empresa

Una empresa presenta los siguientes datos:

ingresos = 125000
gastos = 87000
inversion = 15000

Calcula:
1. Beneficio neto (ingresos menos gastos).
2. Margen de beneficio en porcentaje.
3. ROI (retorno de la inversión), utilizando el beneficio neto como retorno.
4. Porcentaje de gastos sobre ingresos.
5. Indica si la empresa es rentable (beneficio positivo).

```python
ingresos = 125000
gastos = 87000
inversion = 15000

beneficio_neto = ingresos - gastos
margen_beneficio = (beneficio_neto / ingresos) * 100
roi = (beneficio_neto / inversion) * 100
porcentaje_gastos = (gastos / ingresos) * 100
empresa_rentable = beneficio_neto > 0

print("===== ANÁLISIS FINANCIERO =====")
print(f"Beneficio neto: {beneficio_neto} €") # Beneficio neto de la empresa
print(f"Margen de beneficio: {margen_beneficio:.2f}%") # Margen de beneficio de la empresa
print(f"ROI (Retorno de la inversión): {roi:.2f}%") # Retorno de la inversión
print(f"Porcentaje de gastos sobre ingresos: {porcentaje_gastos:.2f}%") # Porcentaje de gastos sobre ingresos
print(f"¿La empresa es rentable?: {'Sí' if empresa_rentable else 'No'}") # Indica si la empresa es rentable o no
```

### Ejercicio 1.3 – Indicadores de clientes

Una tienda online tiene:

clientes_totales = 1200
clientes_nuevos = 180
clientes_perdidos = 75

Calcula:
1. Clientes al final del periodo.
2. Tasa de captación de clientes sobre la base inicial.
3. Tasa de pérdida de clientes sobre la base inicial.
4. Crecimiento neto de clientes en porcentaje.

Considera que los 1.200 clientes son la base inicial.

```python
clientes_totales = 1200
clientes_nuevos = 180
clientes_perdidos = 75

clientes_totales_final = clientes_totales + clientes_nuevos - clientes_perdidos
tasa_captura = (clientes_nuevos / clientes_totales) * 100
tasa_perdidas = (clientes_perdidos / clientes_totales) * 100
crecimiento_neto_porcentaje = (clientes_nuevos - clientes_perdidos) / clientes_totales * 100

print("===== ANÁLISIS DE CLIENTES =====")
print(f"Clientes totales al final del periodo: {clientes_totales_final}") # Clientes totales al final del periodo
print(f"Tasa de captura de nuevos clientes: {tasa_captura:.2f}%") # Tasa de captura de nuevos clientes
print(f"Tasa de pérdidas de clientes: {tasa_perdidas:.2f}%") # Tasa de pérdidas de clientes
print(f"Crecimiento neto de clientes en porcentaje: {crecimiento_neto_porcentaje:.2f}%") # Crecimiento neto de clientes en porcentaje
```

### Ejercicio 2.1 – Clasificación de rendimiento académico

Dispones de una nota de un alumno:

nota = 7.8

Crea un programa que clasifique el resultado:

- Menos de 5: Suspenso.
- Entre 5 y 6: Aprobado.
- Entre 6 y 7: Bien.
- Entre 7 y 9: Notable.
- Entre 9 y 10: Sobresaliente.
- 10: Matrícula de honor.

Añade una validación para impedir notas inferiores a 0 o superiores a 10.

```python
nota = 7.8

if 0 <= nota < 5:
    print("Suspenso")
elif 5 <= nota < 6:
    print("Aprobado")
elif 6 <= nota < 7:
    print("Bien")
elif 7 <= nota < 9:
    print("Notable")
elif 9 <= nota <= 10:
    print("Sobresaliente")
elif nota == 10:
    print("Matrícula de honor")
else:
    print("Nota inválida")
```

### Ejercicio 2.2 – Análisis de ventas mensuales

Una empresa ha registrado estas ventas:

ventas = [1200, 1500, 900, 1800, 2100, 1700,
          800, 2400, 1900, 2200, 2600, 3000]

Utilizando bucles:
1. Calcula las ventas totales.
2. Calcula la media mensual.
3. Cuenta cuántos meses superaron los 2.000 €.
4. Identifica cuántos meses quedaron por debajo de la media.
5. Muestra los meses que superaron la media.

```python
ventas = [1200, 1500, 900, 1800, 2100, 1700,
          800, 2400, 1900, 2200, 2600, 3000]

total_ventas = 0

for venta in ventas:
    total_ventas += venta

media_ventas = total_ventas / len(ventas)

meses_mas_2000 = 0

for venta in ventas:
    if venta > 2000:
        meses_mas_2000 += 1

meses_debajo_media = 0

for venta in ventas:
    if venta < media_ventas:
        meses_debajo_media += 1

print("===== ANÁLISIS DE VENTAS =====")
print(f"Total de ventas: {total_ventas} €") # Total de ventas
print(f"Media de ventas: {media_ventas:.2f} €") # Media de ventas
print(f"Número de meses con ventas superiores a 2000 €: {meses_mas_2000}") # Número de meses con ventas superiores a 2000 €
print(f"Número de meses con ventas por debajo de la media: {meses_debajo_media}") # Número de meses con ventas por debajo de la media

print("Meses con ventas superiores a 2500 €:")

for venta in ventas:
    if venta > 2500:
        print(f"{venta} €")
```

### Ejercicio 2.3 – Detección de anomalías

Una máquina industrial registra las siguientes temperaturas:

temperaturas = [45, 47, 49, 48, 52, 85, 51, 47, 92, 50, 48]

Considera que cualquier temperatura superior a 70 °C es una anomalía.

Tu programa debe:
1. Recorrer todos los registros.
2. Identificar las temperaturas anómalas.
3. Contabilizar el número de anomalías.
4. Calcular el porcentaje de anomalías respecto al total.
5. Mostrar un mensaje de alerta si hay más de una anomalía.

```python
temperaturas = [45, 47, 49, 48, 52, 85, 51, 47, 92, 50, 48]

temperaturas_anomalas = 0

for i in temperaturas:
    if i > 70:
        print(f"Temperatura {i} °C es alta")
        temperaturas_anomalas += 1

if temperaturas_anomalas > 1:
    print(f"Se han detectado {temperaturas_anomalas} temperaturas anómalas")

porcentaje_anomalias = (temperaturas_anomalas / len(temperaturas)) * 100
print(f"Porcentaje de temperaturas anómalas: {porcentaje_anomalias:.2f}%")
```

### Ejercicio 2.4 – Simulación de crecimiento

Una empresa comienza el año con 500 clientes y cada mes aumenta su cartera un 5 %.

Mediante un bucle, simula los próximos 12 meses.

Muestra:

- Número de clientes al finalizar cada mes.
- Número total de clientes al finalizar el año.
- Crecimiento acumulado respecto al inicio.

Redondea los resultados a dos decimales.

```python
clientes_iniciales = 500
clientes = clientes_iniciales
crecimiento_mensual = 0.05

print("===== SIMULACIÓN DE CRECIMIENTO =====")

for mes in range(1, 13):
    clientes = clientes * (1 + crecimiento_mensual)
    print(f"Mes {mes}: {clientes:.2f} clientes")

# Crecimiento acumulado

crecimiento_total = ((clientes - clientes_iniciales) / clientes_iniciales) * 100

print("===== RESULTADOS FINALES =====")
print(f"Clientes iniciales: {clientes_iniciales}")
print(f"Clientes finales {clientes:.2f}")
print(f"Crecimiento acumulado: {crecimiento_total:.2f}%")
```

### Ejercicio 3.1 – Estadísticas de ventas

Dispones de:

ventas = [250, 400, 180, 520, 350, 600, 290, 450, 700, 310]

Calcula sin utilizar librerías externas:
1. Total de ventas.
2. Media.
3. Máximo y mínimo.
4. Segundo valor más alto.
5. Número de ventas superiores a 400.
6. Ordena los valores de menor a mayor.
7. Crea una nueva lista con las ventas superiores a la media.

```python
ventas = [250, 400, 180, 520, 350, 600, 290, 450, 700, 310]

total_ventas = 0
ventas_superiores_400 = 0

for venta in ventas:
    total_ventas += venta
    if venta > 400:
        ventas_superiores_400 += 1

media_ventas = total_ventas / len(ventas)

mayor_venta = max(ventas)

minima_venta = min(ventas)

segunda_mayor_venta = sorted(ventas, reverse=True)[1]

menor_a_mayor = sorted(ventas)

lista_ventas_superiores_media = [venta for venta in ventas if venta > media_ventas]

print("===== ANÁLISIS DE VENTAS =====")
print(f"Total de ventas: {total_ventas} €") # Total de ventas
print(f"Media de ventas: {media_ventas:.2f} €") # Media de ventas
print(f"Número de ventas superiores a 400 €: {ventas_superiores_400}") # Número de ventas superiores a 400 €
print(f"Mayor venta: {mayor_venta} €") # Mayor venta
print(f"Segunda mayor venta: {segunda_mayor_venta} €") # Segunda mayor venta
print(f"Menor venta: {minima_venta} €") # Menor venta
print(f"Ventas ordenadas de menor a mayor: {menor_a_mayor}") # Ventas ordenadas de menor a mayor
print(f"Ventas superiores a la media: {lista_ventas_superiores_media}") # Ventas superiores a la media
```

### Ejercicio 3.2 – Análisis de puntuaciones

Una plataforma recoge las siguientes puntuaciones de sus usuarios:

puntuaciones = [8, 9, 5, 7, 10, 6, 8, 4, 9, 7, 10, 8]

Realiza las siguientes operaciones:
1. Calcula la media.
2. Cuenta cuántos usuarios han obtenido un 10.
3. Crea una lista con las puntuaciones aprobadas (mayores o iguales a 5).
4. Calcula la media de aprobados.
5. Ordena las puntuaciones.
6. Obtén los tres valores más altos.

```python
puntuaciones = [8, 9, 5, 7, 10, 6, 8, 4, 9, 7, 10, 8]

media_puntuaciones = sum(puntuaciones) / len(puntuaciones)
usuarios_10 = 0

for puntuacion in puntuaciones:
    if puntuacion == 10:
        usuarios_10 += 1

lista_puntuaciones_aprobadas = [puntuacion for puntuacion in puntuaciones if puntuacion >= 5]

media_aprobados = sum(lista_puntuaciones_aprobadas) / len(lista_puntuaciones_aprobadas)

puntuaciones_ordenadas = sorted(puntuaciones)

tres_mejores_puntuaciones = sorted(puntuaciones, reverse=True)[:3]

print("===== ANÁLISIS DE PUNTUACIONES =====")
print(f"Media de puntuaciones: {media_puntuaciones:.2f}") # Media de puntuaciones
print(f"Número de usuarios que han dado un 10: {usuarios_10}") # Número de usuarios que han dado un 10
print(f"Puntuaciones aprobadas: {lista_puntuaciones_aprobadas}") # Puntuaciones aprobadas
print(f"Media de puntuaciones aprobadas: {media_aprobados:.2f}") # Media de puntuaciones aprobadas
print(f"Puntuaciones ordenadas de menor a mayor: {puntuaciones_ordenadas}") # Puntuaciones ordenadas de menor a mayor
print(f"Las tres mejores puntuaciones: {tres_mejores_puntuaciones}") # Las tres mejores puntuaciones
```

### Ejercicio 3.3 – Registro de datos mediante tuplas

Una empresa almacena información de sus empleados utilizando tuplas:

empleados = [
    ("Ana", 28, 25000),
    ("Carlos", 35, 32000),
    ("Lucía", 42, 41000),
    ("Pedro", 24, 22000),
    ("Marta", 31, 29000)
]

Cada tupla contiene (nombre, edad, salario).

Calcula:
1. Edad media de los empleados.
2. Salario medio.
3. Empleado con mayor salario.
4. Empleado más joven.
5. Número de empleados que cobran más de 30.000 €.

```python
empleados = [
    ("Ana", 28, 25000),
    ("Carlos", 35, 32000),
    ("Lucía", 42, 41000),
    ("Pedro", 24, 22000),
    ("Marta", 31, 29000)
]

edad_media = sum(empleado[1] for empleado in empleados) / len(empleados)

salario_medio = sum(empleado[2] for empleado in empleados) / len(empleados)

empleado_mayor_salario = max(empleados, key=lambda x: x[2])

empleado_mas_joven = min(empleados, key=lambda x: x[1])

empleados_mas_30000 = [empleado for empleado in empleados if empleado[2] > 30000]

print("===== ANÁLISIS DE EMPLEADOS =====")
print(f"Edad media de los empleados: {edad_media:.2f} años") # Edad media de los empleados
print(f"Salario medio de los empleados: {salario_medio:.2f} €") # Salario medio de los empleados
print(f"Empleado con mayor salario: {empleado_mayor_salario[0]} con {empleado_mayor_salario[2]} €") # Empleado con mayor salario
print(f"Empleado más joven: {empleado_mas_joven[0]} con {empleado_mas_joven[1]} años") # Empleado más joven
print(f"Empleados con salario superior a 30000 €: {[empleado[0] for empleado in empleados_mas_30000]}") # Empleados con salario superior a 30000 €
```
