## Ejercicio 6 (intermedio): conversión de unidades

1. Una temperatura corporal de `36.6 °C` en Fahrenheit (`F = C * 9 / 5 + 32`).
2. Una maratón (`42.195 km`) en millas (1 milla = `1.609344 km`), con 2 decimales.
3. Una temperatura de `451 °F` en Celsius (`C = (F - 32) * 5 / 9`), con 1 decimal.
   Escríbela también sin paréntesis y explica en un comentario por qué el resultado cambia.

```python
# Tu código aquí

# 1. Una temperatura corporal de `36.6 °C` en Fahrenheit (`F = C * 9 / 5 + 32`)
celsius = 36.6
fahrenheit_1 = celsius * 9 / 5 + 32

# 2. Una maratón (42.195 km) a millas (1 milla = 1.609344 km) con 2 decimales
km = 42.195
km_a_millas = 1.609344
millas = km / km_a_millas

# 3. Una temperatura de `451 °F` en Celsius (`C = (F - 32) * 5 / 9`), con 1 decimal
fahrenheit_2 = 451
celsius_correcto = (fahrenheit_2 - 32) * 5 / 9

# 4. Temperatura de 451 °F a Celsius SIN paréntesis
# Debido a la jerarquía de operadores (precedencia), la multiplicación (*) y la división (/)
# se ejecutan antes que la resta (-). Sin paréntesis, Python calcula primero (32 * 5 / 9) 
# y luego resta ese resultado de 451, en lugar de restar 32 al 451 primero. 
# Esto cambia por completo el orden matemático y arroja un valor erróneo.
celsius_sin_parentesis = fahrenheit_2 - 32 * 5 / 9

# --- Mostrar los resultados formateados ---
print(f"1. 36.6 °C equivalen a {fahrenheit_1:.2f} °F")
print(f"2. Una maratón de 42.195 km son {millas:.2f} millas")
print(f"3. 451 °F equivalen a {celsius_correcto:.1f} °C (con paréntesis)")
print(f"4. 451 °F sin paréntesis da {celsius_sin_parentesis:.1f} °C (resultado incorrecto por orden de operadores)")
```

## Ejercicio 7 (intermedio): calculadora de IMC

El índice de masa corporal se calcula como `peso (kg) / altura (m) ** 2`.

1. Con `peso = 72` kg y `altura_cm = 178`, calcula el IMC y muéstralo con 2 decimales.
2. Calcula tres booleanos **sin usar `if`**:
   - `bajo_peso`: IMC menor que 18.5,
   - `peso_normal`: IMC entre 18.5 (incluido) y 25 (excluido),
   - `sobrepeso`: IMC de 25 o más.
3. Comprueba que exactamente uno de los tres es `True` sumándolos.
4. Opcional: pide el peso y la altura con `input()`.

```python
peso = 72
altura_cm = 178

# 1. Con `peso = 72` kg y `altura_cm = 178`, calcula el IMC y muéstralo con 2 decimales

peso = 72

altura_cm = 178

altura_m = altura_cm / 100

imc = peso / altura_m ** 2

print(f"IMC: {imc:.2f}")

# 2. Calcula tres booleanos **sin usar `if`**

bajo_peso = imc < 18.5

peso_normal = 18.5 <= imc < 25

sobrepeso = imc >= 25

print(f"Bajo peso: {bajo_peso}")

print(f"Peso normal: {peso_normal}")

print(f"Sobrepeso: {sobrepeso}")

# 3. Comprueba que exactamente uno de los tres es `True` sumándolos

cantidad_true = bajo_peso + peso_normal + sobrepeso

print(f"Cantidad de True: {cantidad_true}")

print(f"¿Exactamente uno es True? {cantidad_true == 1}")
```

## Ejercicio 8 (intermedio): entrada al cine y descuento (sin condicionales)

El precio de la entrada del cine es **14 €**. Hay un **descuento del 50 %** si se cumple
alguna de estas condiciones:

- la persona es menor de 12 años,
- es mayor de 65 años,
- o tiene **tarjeta de cliente frecuente**.

Datos: `edad` y `tiene_tarjeta` (`True` o `False`).

1. Calcula si le corresponde el descuento (`True` o `False`).
2. Calcula el precio final multiplicando el precio por `0.5` si hay descuento o por `1` si
   no lo hay, **sin usar `if`** (pista: los booleanos valen `1` y `0`).
3. Prueba con varias personas: 8 años sin tarjeta, 30 años sin tarjeta, 30 años con
   tarjeta y 70 años sin tarjeta.
4. Opcional: pide la edad y si tiene tarjeta (`si`/`no`) con `input()`.

Ejemplo de salida: `¿Tiene descuento? True. Precio final: 7.00 €`

```python
# Precio base de la entrada del cine
PRECIO_BASE = 14.0

# --- CASO 1: 8 años sin tarjeta ---
edad_1 = 8
tarjeta_1 = False
tiene_descuento_1 = (edad_1 < 12) or (edad_1 > 65) or tarjeta_1
precio_final_1 = PRECIO_BASE * (1.0 - (0.5 * tiene_descuento_1))

print(f"Perfil (8 años sin tarjeta): ¿Tiene descuento? {tiene_descuento_1}. Precio final: {precio_final_1:.2f} €")

# --- CASO 2: 30 años sin tarjeta ---
edad_2 = 30
tarjeta_2 = False
tiene_descuento_2 = (edad_2 < 12) or (edad_2 > 65) or tarjeta_2
precio_final_2 = PRECIO_BASE * (1.0 - (0.5 * tiene_descuento_2))

print(f"Perfil (30 años sin tarjeta): ¿Tiene descuento? {tiene_descuento_2}. Precio final: {precio_final_2:.2f} €")

# --- CASO 3: 30 años con tarjeta ---
edad_3 = 30
tarjeta_3 = True
tiene_descuento_3 = (edad_3 < 12) or (edad_3 > 65) or tarjeta_3
precio_final_3 = PRECIO_BASE * (1.0 - (0.5 * tiene_descuento_3))

print(f"Perfil (30 años con tarjeta): ¿Tiene descuento? {tiene_descuento_3}. Precio final: {precio_final_3:.2f} €")

# --- CASO 4: 70 años sin tarjeta ---
edad_4 = 70
tarjeta_4 = False
tiene_descuento_4 = (edad_4 < 12) or (edad_4 > 65) or tarjeta_4
precio_final_4 = PRECIO_BASE * (1.0 - (0.5 * tiene_descuento_4))

print(f"Perfil (70 años sin tarjeta): ¿Tiene descuento? {tiene_descuento_4}. Precio final: {precio_final_4:.2f} €")
```

## Ejercicio 9 (integrador): ticket de compra con IVA

Una tienda online vende estos artículos (precios **sin IVA**):

| Artículo | Precio unitario | Cantidad |
|---|---|---|
| Camiseta | 19.99 € | 2 |
| Pantalón | 39.95 € | 1 |
| Calcetines | 4.50 € | 3 |

Reglas de la tienda:

- **Descuento del 10 %** sobre el subtotal si el subtotal es de 75 € o más, **o** si el
  cliente es socio.
- **Gastos de envío de 4.95 €** si, tras el descuento, la compra es inferior a 50 €; si no,
  el envío es gratis.
- Se aplica un **IVA del 21 %** sobre la base (subtotal - descuento + envío).

Calcula, **sin usar `if`**, el subtotal, el descuento, el envío, la base, el IVA y el total,
y muestra un ticket con todos los importes a 2 decimales. Usa constantes para el IVA, el
porcentaje de descuento y el coste de envío. Después cambia `es_socio` y las cantidades para
comprobar que las reglas funcionan (por ejemplo, 1 camiseta y nada más).

```python
# --- CONSTANTES ---
IVA_PORCENTAJE = 0.21
DESCUENTO_PORCENTAJE = 0.10
UMBRAL_DESCUENTO = 75.0
COSTE_ENVIO = 4.95
UMBRAL_ENVIO = 50.0

# --- DATOS DE ENTRADA ---
precio_camiseta = 19.99
cantidad_camiseta = 2

precio_pantalon = 39.95
cantidad_pantalon = 1

precio_calcetines = 4.50
cantidad_calcetines = 3

es_socio = False  # Cambiar a False para comprobar

# --- 1. CÁLCULO DEL SUBTOTAL ---
subtotal = (precio_camiseta * cantidad_camiseta) + \
           (precio_pantalon * cantidad_pantalon) + \
           (precio_calcetines * cantidad_calcetines)

# --- 2. CÁLCULO DEL DESCUENTO (10% si subtotal >= 75 o es_socio) ---
# Comprobamos si se cumple la condición de descuento (devuelve True o False)
aplica_descuento = (subtotal >= UMBRAL_DESCUENTO) or es_socio

# El importe del descuento se obtiene multiplicando el subtotal por el porcentaje y por el booleano (1 o 0)
importe_descuento = subtotal * DESCUENTO_PORCENTAJE * aplica_descuento
subtotal_con_descuento = subtotal - importe_descuento

# --- 3. CÁLCULO DE LOS GASTOS DE ENVÍO (4.95 € si subtotal con descuento < 50) ---
# Comprobamos si el importe tras descuento es menor de 50
es_menor_50 = subtotal_con_descuento < UMBRAL_ENVIO

# Si es menor de 50, se multiplica por 1 (suma 4.95 €); si no, por 0 (envío gratis)
gastos_envio = COSTE_ENVIO * es_menor_50

# --- 4. CÁLCULO DE LA BASE, IVA Y TOTAL ---
base_imponible = subtotal_con_descuento + gastos_envio
importe_iva = base_imponible * IVA_PORCENTAJE
total = base_imponible + importe_iva

# --- 5. MOSTRAR EL TICKET ---
print("========================================")
print("             TICKET DE COMPRA           ")
print("========================================")
print(f"Camisetas ({cantidad_camiseta}x): {(precio_camiseta * cantidad_camiseta):6.2f} €")
print(f"Pantalones ({cantidad_pantalon}x): {(precio_pantalon * cantidad_pantalon):6.2f} €")
print(f"Calcetines ({cantidad_calcetines}x): {(precio_calcetines * cantidad_calcetines):6.2f} €")
print("----------------------------------------")
print(f"Subtotal: {subtotal:6.2f} €")
print(f"Descuento ({int(DESCUENTO_PORCENTAJE * 100)}%): -{importe_descuento:6.2f} €")
print(f"Gastos de envío: {gastos_envio:6.2f} €")
print("----------------------------------------")
print(f"Base imponible: {base_imponible:6.2f} €")
print(f"IVA ({int(IVA_PORCENTAJE * 100)}%): {importe_iva:6.2f} €")
print("========================================")
print(f"TOTAL A PAGAR: {total:6.2f} €")
print("========================================")
```
