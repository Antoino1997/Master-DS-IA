## Ejercicio 1 (básico): ficha de un registro

Una fila de un dataset de clientes contiene estos datos:

- identificador: `1043`
- nombre: `Lucía Martín`
- saldo: `2350.80`
- cliente activo: sí
- fecha de baja: no tiene (dato ausente)

Crea una variable por campo, con nombres en `snake_case` y el tipo adecuado, y muestra cada
valor junto con su tipo usando f-strings.

Pista: no llames `id` a la variable del identificador: `id` es una función de Python.

````python
identificador = 1043
nombre = "Lucía Martín"
saldo = 2350.80
cliente_activo = True
fecha_de_baja = None

print(f"Identificador: {identificador} — Tipo: {type(identificador)}")
print(f"Nombre: {nombre} — Tipo: {type(nombre)}")
print(f"Saldo: {saldo} — Tipo: {type(saldo)}")
print(f"Cliente activo: {cliente_activo} — Tipo: {type(cliente_activo)}")
print(f"Fecha de baja: {fecha_de_baja} — Tipo: {type(fecha_de_baja)}")
````

## Ejercicio 2 (básico): predice el tipo

Antes de ejecutar nada, escribe en un comentario el **valor y el tipo** que crees que dará
cada expresión. Después compruébalo con `print()` y `type()`.

1. `7 / 2`
2. `8 / 4`
3. `3 * 1.0`
4. `"7" + "2"`
5. `int("42")`
6. `str(4.5)`
7. `float(3)`
8. `int(7.9)`

````python
print(7 / 2, type(7 / 2)) # Float
print(8 // 2, type(8 // 2)) # Integer
print(3 * 1.0, type(3 * 1.0)) # Float
print("7" + "2", type("7" + "2")) # String
print(int("42"), type(int("42"))) # Integer
print(str(4.5), type(str(4.5))) # String
print(float(3), type(float(3))) # Float
print(int(7.9), type(int(7.9))) # Integer
````

## Ejercicio 6 (intermedio): la precisión de los decimales en un carrito

Un carrito contiene un artículo de `0.10 €` y otro de `0.20 €`.

1. Calcula el total y compáralo con `0.3` usando `==`. ¿Qué ocurre?
2. Repite la comparación redondeando antes a 2 decimales.
3. Otro carrito tiene 3 unidades de un artículo de `19.99 €`. Muestra el total tal cual,
   redondeado con `round()` y formateado con una f-string a 2 decimales.

````python
carrito_1 = 0.10
carrito_2 = 0.20

total_carrito = carrito_1 + carrito_2
print(total_carrito == 0.3)

total_carrito_bien = round(total_carrito, 2)
print(total_carrito_bien == 0.3)

cantidad = 3
precio = 19.99

carrito_3 = round(cantidad * precio, 2)
print(f"{carrito_3:.2f}")
````

## Ejercicio 7 (integrador): formulario de alta

Escribe un pequeño formulario que pida por `input()`:

- el nombre,
- el año de nacimiento,
- la altura en centímetros,
- si acepta recibir la newsletter (`s` o `n`).

Después:

1. Convierte cada dato al tipo adecuado.
2. Calcula la edad aproximada (año actual menos año de nacimiento; guarda el año actual en
   una constante `ANIO_ACTUAL`) y la altura en metros.
3. Calcula `acepta_newsletter` como `True`/`False` **sin usar `if`**.
4. Muestra un resumen con f-strings que incluya el tipo de cada dato convertido.

Ejemplo de salida para `Marta`, `1998`, `165`, `s`:

```
Nombre: Marta (str)
Edad aproximada: 28 años (int)
Altura: 1.65 m (float)
Acepta newsletter: True (bool)
```

````python
AÑO_ACTUAL = 2026

# 1. Solicitar los datos
nombre = input("Introduce tu nombre: ")
año_nacimiento = input("Introduce tu año de nacimiento: ")
altura_cm = input("Introduce tu altura en centímetros: ")
newsletter = input("¿Aceptas recibir la newsletter? (s/n): ")

# 2. Convertir los datos a sus tipos correspondientes
año_nacimiento = int(año_nacimiento)
altura_cm = float(altura_cm)

# 3. Calcular edad y altura
edad = AÑO_ACTUAL - año_nacimiento
altura_m = altura_cm / 100

# 4. Convertir la respuesta de newsletter a booleano
acepta_newsletter = newsletter.lower() == "s"

# 5. Mostrar el resumen
print("\n===== RESUMEN DEL REGISTRO =====")
print(f"Nombre: {nombre} ({type(nombre).__name__})")
print(f"Edad aproximada: {edad} años ({type(edad).__name__})")
print(f"Altura: {altura_m:.2f} m ({type(altura_m).__name__})")
print(f"Acepta newsletter: {acepta_newsletter} ({type(acepta_newsletter).__name__})")
````
